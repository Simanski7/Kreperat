```python
from IPython.display import HTML
import colorsys

# vygenerujeme 720 barev
colors = [
    '#' + ''.join(f'{int(c*255):02x}' for c in colorsys.hsv_to_rgb(i/720, 1, 1))
    for i in range(720)
]

HTML(f"""
<canvas id="turtleCanvas" width="900" height="900" style="background:black"></canvas>

<script>
var canvas = document.getElementById("turtleCanvas");
var ctx = canvas.getContext("2d");

var x = canvas.width / 2;
var y = canvas.height / 2;
var angle = 0;

function forward(d) {{
    var rad = angle * Math.PI/180;
    var nx = x + d * Math.cos(rad);
    var ny = y + d * Math.sin(rad);

    ctx.beginPath();
    ctx.moveTo(x, y);
    ctx.lineTo(nx, ny);
    ctx.stroke();

    x = nx;
    y = ny;
}}

function left(a) {{ angle -= a; }}
function right(a) {{ angle += a; }}
function color(c) {{ ctx.strokeStyle = c; }}
function width(w) {{ ctx.lineWidth = w; }}

width(1.7);
var A = 137.508;

var colors = {colors};

for (let i = 0; i < 2000; i++) {{
    color(colors[i % 720]);
    forward(i * 0.45);
    left(A + i * 0.015);
}}

for (let i = 0; i < 1600; i++) {{
    color(colors[(i*5) % 720]);
    forward(i * 0.3);
    right(A*1.2 + i * 0.01);
}}
</script>
""")
```

```python
from IPython.display import Javascript
import colorsys

# Zlatý úhel
A = 137.507764

# paleta 720 barev
colors = [
    '#' + ''.join(f'{int(c*255):02x}' for c in colorsys.hsv_to_rgb(i/720, 1, 1))
    for i in range(720)
]

js_code = f"""
width(1.8);
var A = {A};
"""

# hlavní spirála
for i in range(2000):
    col = colors[i % 720]
    dist = i * 0.45
    ang = A + (i * 0.015)

    js_code += f"""
color("{col}");
forward({dist});
left({ang});
"""

# druhá fraktální vrstva
for i in range(1600):
    col = colors[(i * 5) % 720]
    dist = i * 0.3
    ang = -A * 1.2 - (i * 0.01)

    js_code += f"""
color("{col}");
forward({dist});
right({ang});
"""

Javascript(js_code)
```

```python
from IPython.display import HTML
import colorsys

# 1600 neonových barev
colors = [
    '#' + ''.join(f'{int(c*255):02x}' for c in colorsys.hsv_to_rgb(i/1600, 1, 1))
    for i in range(1600)
]

HTML(f"""
<canvas id="turtleCanvas" width="1400" height="1400" style="background:black"></canvas>

<script>
var canvas = document.getElementById("turtleCanvas");
var ctx = canvas.getContext("2d");

// --- NEONOVÝ TEXT ---
ctx.save();
ctx.font = "bold 110px Orbitron, sans-serif";
ctx.textAlign = "center";
ctx.fillStyle = "#ff00ff";
ctx.shadowColor = "#ff00ff";
ctx.shadowBlur = 40;
ctx.fillText("HYPERFRACTAL BANGER", canvas.width/2, 180);
ctx.restore();

// --- TURTLE ENGINE ---
var x = canvas.width / 2;
var y = canvas.height / 2;
var angle = 0;

function forward(d) {{
    var rad = angle * Math.PI/180;
    var nx = x + d * Math.cos(rad);
    var ny = y + d * Math.sin(rad);

    ctx.beginPath();
    ctx.moveTo(x, y);
    ctx.lineTo(nx, ny);
    ctx.stroke();

    x = nx;
    y = ny;
}}

function left(a) {{ angle -= a; }}
function right(a) {{ angle += a; }}
function color(c) {{ ctx.strokeStyle = c; }}
function width(w) {{ ctx.lineWidth = w; }}

width(1.5);
var A = 137.508;
var colors = {colors};

//////////////////////////
// VRSTVA 1 – GALAXIE
//////////////////////////
for (let i = 0; i < 3500; i++) {{
    color(colors[(i * 7) % 1600]);
    forward(i * 0.33);
    left(A + Math.sin(i/60) * 12);
}}

//////////////////////////
// VRSTVA 2 – TURBULENCE
//////////////////////////
x = canvas.width / 2;
y = canvas.height / 2;
angle = 0;

for (let i = 0; i < 2800; i++) {{
    color(colors[(i * 19) % 1600]);
    forward(i * 0.24);
    right(A * 0.9 + Math.cos(i/35) * 22);
}}

//////////////////////////
// VRSTVA 3 – CHAOS VORTEX
//////////////////////////
x = canvas.width / 2;
y = canvas.height / 2;
angle = 0;

for (let i = 0; i < 5000; i++) {{
    let noise = Math.sin(i/50) * 40 + Math.cos(i/23) * 25;
    color(colors[(i * 31) % 1600]);
    forward(i * 0.08);
    left(A * 1.5 + noise);
}}
</script>
""")
```
