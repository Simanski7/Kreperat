```python
# Průměr libovolného počtu čísel
def prumer(*cisla: float) -> float | None:
    '''
    Vypočítá a vrátí průměr zadaných čísel.

    Args:
        *cisla (float): Libovolný počet čísel (integerů nebo floatů).

    Returns:
        float | None: Průměr zadaných čísel. Pokud není zadáno žádné číslo, vrátí None.
    '''
    if not cisla:
        return None
    return sum(cisla) / len(cisla)

print(prumer(1, 2, 3, 4)) # Example with numbers
print(prumer())         # Example with no numbers
print(prumer(1.5, 2.5, 3.0)) # Example with floats
```

```python
def vypis(text:str, *text2:str, opakovani: int = 1):
    '''
    Vypíše zadaný text a cyklicky opakuje tisk prvků z `text2`.

    Args:
        text (str): Hlavní text k vypsání.
        *text2 (str): Libovolný počet dalších textů, které budou cyklicky vypsány.
        opakovani (int, optional): Kolikrát se má text vypsat. Výchozí je 1.
    '''
    for i in range(opakovani):
        print(text, end = ", ")
        if text2:
            index = i % len(text2)
            print(text2[index])  # vypíše položku texty2
        else:
            print()
```

```python
def vypis_info(**kwargs) -> None:
  '''
  Vypíše všechny předané klíčové argumenty a jejich hodnoty.

  Args:
    **kwargs: Libovolný počet klíčových argumentů.

  Returns:
    None: Funkce nic nevrací, pouze tiskne na konzoli.
  '''
  for klic, hodnota in kwargs.items():
    print(f"{klic}: {hodnota}")

vypis_info(jmeno="Petr", vek=30, mesto="Praha")
```

Vykoušejte různé způsoby volání funkcí

```python
def mix(a: any, b: any, *args: any, c: int = 0, **kwargs: any):
    '''
    Demonstruje pořadí zpracování různých typů argumentů ve funkci.

    Args:
        a (any): První poziční argument.
        b (any): Druhý poziční argument.
        *args (any): Libovolný počet dalších pozičních argumentů, shromážděných do tuple.
        c (int, optional): Pojmenovaný argument s výchozí hodnotou. Výchozí je 0.
        **kwargs (any): Libovolný počet klíčových argumentů, shromážděných do slovníku.
    '''
    print("a =", a)
    print("b =", b)
    print("args =", args)
    print("c =", c)
    print("kwargs =", kwargs)

mix(1, 2, 3, 4, 5, c=10, x=100, y=200)
```

### Příklady volání funkce `mix`

```python
# 1. Pouze poziční argumenty
print("\n--- Pouze poziční argumenty ---")
mix(10, 20)

# 2. Poziční argumenty a *args
print("\n--- Poziční argumenty a *args ---")
mix(10, 20, 'a', 'b', 'c')

# 3. Poziční argumenty a pojmenovaný argument 'c'
print("\n--- Poziční argumenty a pojmenovaný argument 'c' ---")
mix(10, 20, c=5)

# 4. Poziční argumenty a **kwargs
print("\n--- Poziční argumenty a **kwargs ---")
mix(10, 20, nazev='ukazka', hodnota=123)

# 5. Poziční argumenty, *args, pojmenovaný argument 'c' a **kwargs (úplný příklad)
print("\n--- Úplný příklad ---")
mix(1, 2, 'hello', 4.5, True, c=99, key1='val1', key2=False)
```

Ošetřete dělení nulou nějakým jiným způsobem.

```python
def deleni(a: float, b: float) -> float | None:
    '''
    Provádí dělení dvou čísel. Pokud je dělitel nulový, vrátí None.

    Args:
        a (float): Dělenec.
        b (float): Dělitel.

    Returns:
        float | None: Výsledek dělení, nebo None, pokud je dělitel nulový.
    '''
    if b == 0:
        return None  # Místo vyvolání chyby vrátíme None
    return a / b

print(deleni(10, 2))
print(deleni(10, 0)) # Bude vráceno None
```

### Generování náhodných čísel

```python
import random

print(random.randint(1, 10))     # celé číslo 1–10
print(random.random())           # číslo 0–1
print(random.uniform(5, 15))     # desetinné číslo 5–15
```

### Vysvětlení funkcí pro generování náhodných čísel:

*   **`random.randint(a, b)`**: Vrací náhodné celé číslo N takové, že `a <= N <= b`. Oba koncové body jsou zahrnuty.
*   **`random.random()`**: Vrací náhodné desetinné číslo (float) v rozsahu `0.0 <= x < 1.0`.
*   **`random.uniform(a, b)`**: Vrací náhodné desetinné číslo (float) v rozsahu `a <= x <= b` nebo `b <= x <= a`.


```python
import random

print("--- Generování náhodných čísel ---")
print(f"Náhodné celé číslo 1-10: {random.randint(1, 10)}")
print(f"Náhodné desetinné číslo 0-1: {random.random():.4f}") # Formátování pro lepší čitelnost
print(f"Náhodné desetinné číslo 5-15: {random.uniform(5, 15):.4f}") # Formátování pro lepší čitelnost

print("\n--- Náhodný výběr a zamíchání --- ")
seznam = ['jablko', 'banán', 'třešeň', 'datle']
print(f"Náhodný výběr ze seznamu: {random.choice(seznam)}")

random.shuffle(seznam)
print(f"Zamíchaný seznam: {seznam}")

print("\n--- Náhodný vzorek bez opakování ---")
print(f"Náhodné 2 prvky ze seznamu: {random.sample(seznam, 2)}")
```

Vylepšete funkci  nahodny_prumer (docstring, typové hinty, *args, klíčové parametry)

```python
import random

def nahodny_prumer(*cisla: float, pocet_generovanych: int = 0, min_hodnota: int = 1, max_hodnota: int = 100) -> float | None:
    '''
    Vypočítá průměr zadaných čísel, nebo vygeneruje náhodná čísla a vypočítá jejich průměr.

    Args:
        *cisla (float): Volitelný počet čísel, jejichž průměr se má vypočítat.
                        Pokud jsou čísla zadána, 'pocet_generovanych' je ignorován.
        pocet_generovanych (int): Volitelný. Pokud nejsou zadána žádná 'cisla',
                                  určuje kolik náhodných čísel se má vygenerovat.
                                  Výchozí je 0.
        min_hodnota (int): Minimální hodnota pro generovaná náhodná čísla. Výchozí je 1.
        max_hodnota (int): Maximální hodnota pro generovaná náhodná čísla. Výchozí je 100.

    Returns:
        float | None: Průměr čísel, nebo None, pokud nejsou k dispozici žádná čísla k průměrování.
    '''
    numbers_to_average = list(cisla)

    if not numbers_to_average: # Pokud nejsou předána žádná čísla přes *cisla
        if pocet_generovanych > 0:
            numbers_to_average = [random.randint(min_hodnota, max_hodnota) for _ in range(pocet_generovanych)]
        else:
            return None # Žádná explicitní čísla a žádná žádost o generování náhodných

    if not numbers_to_average: # Zkontrolovat znovu, zda seznam není stále prázdný
        return None

    return sum(numbers_to_average) / len(numbers_to_average)

# Příklady použití:
print("Průměr zadaných čísel:", nahodny_prumer(1, 2, 3, 4, 5))
print("Průměr 5 náhodných čísel (výchozí rozsah):", nahodny_prumer(pocet_generovanych=5))
print("Průměr 3 náhodných čísel v rozsahu 10-20:", nahodny_prumer(pocet_generovanych=3, min_hodnota=10, max_hodnota=20))
print("Prázdné volání (žádná čísla, žádné generování):", nahodny_prumer())
print("Kombinace (zadaná čísla mají přednost):", nahodny_prumer(10, 20, pocet_generovanych=5))
```

```python
import random
import matplotlib.pyplot as plt

# Generujeme 1000 náhodných čísel v rozsahu 1-100
random_numbers = [random.randint(1, 100) for _ in range(1000)]

# Vytvoření histogramu
plt.figure(figsize=(10, 6))
plt.hist(random_numbers, bins=20, edgecolor='black', alpha=0.7)
plt.title('Histogram náhodně generovaných čísel (1-100)')
plt.xlabel('Hodnota čísla')
plt.ylabel('Frekvence')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
