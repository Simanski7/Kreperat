```
import json

def parse_input(text):
    """
    Automatické rozpoznání formátu:
    - řádky
    - čárky
    - odrážky (-, *, •)
    - kombinované formáty
    """

    # Rozdělení podle nových řádků nebo čárek
    if "\n" in text:
        raw_items = text.split("\n")
    else:
        raw_items = text.split(",")

    items = []

    for item in raw_items:
        cleaned = item.strip()

        # Odstranění odrážek
        for bullet in ["- ", "* ", "• ", "– "]:
            if cleaned.startswith(bullet):
                cleaned = cleaned[len(bullet):]

        # Pokud je položka prázdná, přeskočíme
        if cleaned == "":
            continue

        # Pokus o převod na číslo
        try:
            if "." in cleaned:
                cleaned = float(cleaned)
            else:
                cleaned = int(cleaned)
        except ValueError:
            pass  # zůstane textem

        items.append(cleaned)

    return items


def save_to_file(items, filename="vystup.json"):
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(items, f, ensure_ascii=False, indent=4)
    print(f"\nSoubor uložen jako: {filename}")


def main():
    print("========================================")
    print("  Program pro převod textového seznamu")
    print("========================================\n")
    print("Vlož svůj zkopírovaný seznam (může být více řádků).")
    print("Zakonči zadání prázdným řádkem:\n")

    lines = []
    while True:
        line = input()
        if line.strip() == "":
            break
        lines.append(line)

    text = "\n".join(lines)

    print("\n--- Zpracovávám seznam ---\n")
    items = parse_input(text)

    print("Výsledný Python list:\n")
    print(items)

    print("\nChceš uložit tento list do souboru? (a/n)")
    if input("> ").lower().startswith("a"):
        save_to_file(items)

    print("\nHotovo! 👍")


if __name__ == "__main__":
    main()
```
