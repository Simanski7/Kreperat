```python
# Aktivace grafů pro Google Colab
%matplotlib inline

# =========================================
# DATASET: Pražské tramvaje
# =========================================

trams = [
    ("Tatra T1", 60, 1951),
    ("Tatra T2", 65, 1955),
    ("Tatra T3", 65, 1960),
    ("Tatra T3SUCS", 65, 1976),
    ("Tatra T3R.P", 65, 2000),
    ("Tatra T3M", 65, 1976),
    ("Tatra T6A5", 65, 1991),
    ("Tatra KT4D (ex-DDR)", 65, 1976),
    ("Tatra K2", 65, 1966),
    ("Tatra K2S", 65, 2003),
    ("Tatra KT8D5", 65, 1984),
    ("Tatra KT8D5.RN2P", 65, 2005),
    ("Škoda 14T", 60, 2005),
    ("Škoda 15T", 60, 2009),
    ("RT6N1", 60, 1993),
    ("VarioLF", 65, 2003),
    ("VarioLF2", 65, 2009),
    ("Pragoimex EVO1", 50, 2015),
    ("Pragoimex EVO2", 60, 2015)
]

# =========================================
# ŘAZENÍ
# =========================================

sorted_by_year = sorted(trams, key=lambda x: x[2])
sorted_by_speed = sorted(trams, key=lambda x: x[1], reverse=True)

print("=== Seřazeno podle roku výroby ===")
for t in sorted_by_year:
    print(t)

print("\n=== Seřazeno podle rychlosti ===")
for t in sorted_by_speed:
    print(t)

# =========================================
# FILTROVÁNÍ
# =========================================

modern = [t for t in trams if t[2] >= 2000]
fast = [t for t in trams if t[1] >= 65]

print("\n=== Moderní tramvaje (2000+) ===")
for t in modern:
    print(t)

print("\n=== Rychlé tramvaje (>= 65 km/h) ===")
for t in fast:
    print(t)

# =========================================
# POPISNÉ STATISTIKY
# =========================================

import statistics as stats

speeds = [t[1] for t in trams]
years = [t[2] for t in trams]

print("\n=== Statistiky ===")
print("Průměrná rychlost:", stats.mean(speeds))
print("Medián rychlosti:", stats.median(speeds))
print("Nejstarší rok výroby:", min(years))
print("Nejnovější rok výroby:", max(years))

# =========================================
# HISTOGRAM 1: Maximální rychlosti
# =========================================

import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(speeds, bins=5, color='orange', edgecolor='black')
plt.xlabel("Maximální rychlost (km/h)")
plt.ylabel("Počet tramvají")
plt.title("Histogram maximálních rychlostí tramvají")
plt.grid(axis="y", alpha=0.3)
plt.show()

# =========================================
# HISTOGRAM 2: Dekády výroby
# =========================================

decades = [(rok // 10) * 10 for rok in years]

plt.figure(figsize=(8,5))
plt.hist(decades, bins=8, color='green', edgecolor='black')
plt.xlabel("Dekáda výroby")
plt.ylabel("Počet typů")
plt.title("Histogram tramvají podle dekády výroby")
plt.grid(axis="y", alpha=0.3)
plt.show()

# =========================================
# HISTOGRAM 3: Délky názvů tramvají
# =========================================

name_lengths = [len(name) for (name, _, _) in trams]

plt.figure(figsize=(8,5))
plt.hist(name_lengths, bins=10, color='purple', edgecolor='black')
plt.xlabel("Délka názvu (počet znaků)")
plt.ylabel("Počet typů")
plt.title("Histogram délek názvů tramvají")
plt.grid(axis="y", alpha=0.3)
plt.show()
```
